```python
def run_query(sql, description=""):
  """
  Runs a SQL query and returns result as a Pandas DataFrame

  Parameters:
  -------------
  sql          :str - SQL query string to execute
  description  :str - A label to print above the results

  Returns:
  ----------
   Pandas DataFrame containing the query results
  """
  if description:
    print(f"\n{'='*55}")
    print(f"{description}")
    print(f"{'=*55'}")

  result=pd.read_sql_query(sql, conn)
  # 1.Executes the SQL query against the database(conn)
  # 2.Converts the result into Pandas DataFrame
  print(result.to_string(index=False))
  return result
```

```python
query1="""
SELECT student_id,name,department,math_score,attendance_percentage
FROM students
LIMIT 10
"""
result1=run_query(query1,"Query 1:First 10 Studnets (SELECT + LIMIT)")
```

```python
query2="""
SELECT name,department,math_score
FROM students
ORDER BY math_score DESC
LIMIT 10
"""
result2=run_query(query2,"Query 2:Top 10 Students (ORDER BY + LIMIT)")
```

```python
query3="""
SELECT name,math_score,science_score,programming_score,attendance_percentage
FROM students
WHERE department='Computer Science'
ORDER BY programming_score DESC
"""
res3=run_query(query3,"Query 3:Computer Science Students (WHERE filter)")
print(f"\nTotal cs studnets found:{len(res3)}")
```

```python
query4="""
SELECT name,department,attendance_percentage
FROM students
WHERE attendance_percentage>90
AND department != 'Civil'
ORDER BY attendance_percentage DESC
"""
res4=run_query(query4)
```

```python
query5="""
SELECT
department,
COUNT(*) AS num_students,
ROUND(AVG(math_score),2) AS avg_math,
ROUND(AVG(science_score),2) AS avg_science,
ROUND(AVG(programming_score),2) AS avg_programming,
ROUND(AVG(attendance_percentage),2) AS avg_attendance
FROM students
GROUP BY department
ORDER BY num_students DESC
"""
res5=run_query(query5)
```

```python
query6="""
SELECT
department,
ROUND(AVG(math_score),2) AS avg_math,
COUNT(*) AS students
FROM students
GROUP BY department
HAVING AVG(math_score)>70
ORDER BY avg_math DESC
"""
res6=run_query(query6)
```

```python
query7="""
SELECT
gender,
COUNT(*) AS num_students,
ROUND(AVG(math_score),2) AS avg_math,
ROUND(AVG(programming_score),2) AS avg_programming,
ROUND(AVG(attendance_percentage),2) AS avg_attendance
FROM students
GROUP BY gender
"""
res7=run_query(query7)
```

```python
query8="""
SELECT
name,
department,
math_score+science_score+english_score+programming_score AS total_score,
attendance_percentage
FROM students
ORDER BY total_score DESC
LIMIT 10
"""
res8=run_query(query8)
```

```python
import sqlite3
import pandas as pd # Ensure pandas is imported

# Unconditionally re-load df to ensure it's the correct student performance data
df = pd.read_csv('/content/student_performance.csv')

# Create an in-memory SQLite database
conn = sqlite3.connect(':memory:')

# Save the DataFrame to a SQLite table named 'students'
df.to_sql('students', conn, if_exists='replace', index=False)

print("DataFrame successfully loaded into an in-memory SQLite database.")
```

```python
import pandas as pd # Ensure pandas is imported
query_1 = """
SELECT
    department,
    ROUND(AVG(math_score), 1) AS average_math_score
FROM
    students
GROUP BY
    department
ORDER BY
    average_math_score DESC;
"""
avg_math_score_df = pd.read_sql_query(query_1, conn)
print("Average math score for each department:")
display(avg_math_score_df)
```

```python
import pandas as pd # Ensure pandas is imported
query_2 = """
SELECT
    department,
    COUNT(*) AS number_of_students
FROM
    students
GROUP BY
    department;
"""
students_per_department_df = pd.read_sql_query(query_2, conn)
print("Number of students in each department:")
display(students_per_department_df)
```

```python
import pandas as pd # Ensure pandas is imported
query_3 = """
SELECT
    name,
    (math_score + science_score + english_score + programming_score) AS total_score
FROM
    students
ORDER BY
    total_score DESC
LIMIT 8;
"""
top_students_df = pd.read_sql_query(query_3, conn)
print("Top 8 students based on total marks:")
display(top_students_df)
```

```python
import pandas as pd # Ensure pandas is imported
query_4 = """
SELECT
    gender,
    ROUND(AVG(attendance_percentage), 1) AS average_attendance_percentage
FROM
    students
GROUP BY
    gender;
"""
avg_attendance_df = pd.read_sql_query(query_4, conn)
print("Average attendance percentage by gender:")
display(avg_attendance_df)
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Create the figure and subplots
fig, axes = plt.subplots(2, 2, figsize=(18, 16))
fig.suptitle('Student Performance Dashboard', fontsize=24, fontweight='bold', y=1.02)

# --- Panel 1: Vertical bar chart showing average math score by department ---
ax1 = axes[0, 0]
bars1 = ax1.bar(avg_math_score_df['department'], avg_math_score_df['average_math_score'], color=['skyblue', 'lightcoral', 'lightgreen'])
ax1.set_title('Average Math Score by Department', fontsize=16)
ax1.set_xlabel('Department', fontsize=12)
ax1.set_ylabel('Average Math Score', fontsize=12)
ax1.set_ylim(0, 105)
ax1.grid(axis='y', linestyle='--', alpha=0.7)
# Add value labels
for bar in bars1:
    yval = bar.get_height()
    ax1.text(bar.get_x() + bar.get_width()/2, yval + 1, round(yval, 1), ha='center', va='bottom', fontsize=10)

# --- Panel 2: Pie chart showing the number of students in each department ---
ax2 = axes[0, 1]
colors = ['#ff9999', '#66b3ff', '#99ff99', '#ffcc99'] # Custom colors
explode = [0.05] * len(students_per_department_df) # Explode slices slightly

wedges, texts, autotexts = ax2.pie(
    students_per_department_df['number_of_students'],
    labels=students_per_department_df['department'],
    autopct=lambda p: '{:.1f}%'.format(p),
    colors=colors,
    explode=explode,
    pctdistance=0.85, # Distance of percentage labels from the center
    startangle=90,
    wedgeprops={'linewidth': 1, 'edgecolor': 'white'}
)
ax2.set_title('Number of Students per Department', fontsize=16)
# Equal aspect ratio ensures that pie is drawn as a circle.
ax2.axis('equal')
# Customize percentage labels
for autotext in autotexts:
    autotext.set_color('black')
    autotext.set_fontsize(10)

# --- Panel 3: Horizontal bar chart showing the Top 8 students based on total score ---
ax3 = axes[1, 0]
# Sort for highest scorer at the top
top_students_df_sorted = top_students_df.sort_values(by='total_score', ascending=True)
bars3 = ax3.barh(
    top_students_df_sorted['name'],
    top_students_df_sorted['total_score'],
    color=plt.cm.viridis(np.linspace(0, 1, len(top_students_df_sorted)))
)
ax3.set_title('Top 8 Students by Total Score', fontsize=16)
ax3.set_xlabel('Total Score', fontsize=12)
ax3.set_ylabel('Student Name', fontsize=12)
ax3.grid(axis='x', linestyle='--', alpha=0.7)
# Add value labels inside bars
for bar in bars3:
    xval = bar.get_width()
    ax3.text(xval - 5, bar.get_y() + bar.get_height()/2, int(xval), ha='right', va='center', color='white', fontsize=10)

# --- Panel 4: Bar chart showing average attendance percentage by gender ---
ax4 = axes[1, 1]
gender_colors = {'Male': 'blue', 'Female': 'red', 'Other': 'green'}
# Ensure the colors match the actual gender categories in your data
bar_colors = [gender_colors.get(g, 'gray') for g in avg_attendance_df['gender']]
bars4 = ax4.bar(
    avg_attendance_df['gender'],
    avg_attendance_df['average_attendance_percentage'],
    color=bar_colors
)
ax4.set_title('Average Attendance Percentage by Gender', fontsize=16)
ax4.set_xlabel('Gender', fontsize=12)
ax4.set_ylabel('Average Attendance %', fontsize=12)
ax4.set_ylim(0, 105)
ax4.grid(axis='y', linestyle='--', alpha=0.7)
# Add percentage labels
for bar in bars4:
    yval = bar.get_height()
    ax4.text(bar.get_x() + bar.get_width()/2, yval + 1, f'{round(yval, 1)}%', ha='center', va='bottom', fontsize=10)

# Adjust layout and save the dashboard
plt.tight_layout(rect=[0, 0.03, 1, 0.98]) # Adjust rect to make space for suptitle
plt.savefig('student_dashboard.png', dpi=150, bbox_inches='tight')

print("Dashboard 'student_dashboard.png' saved successfully!")

# Close the database connection
conn.close()
```
